```python
import pandas as pd
df = pd.read_csv('car_sales_cleaned1.csv')
df
```

```python
df = df.drop_duplicates()
df = df.dropna() # drop rows with null values
df.shape
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
import converter as conv

mapper = conv.columnsMapper(['Gender', 'Company', 'Model', 'Transmission', 'Color'], df)
df_conv = conv.convertX(df, mapper)

X = df_conv[['Model', 'Transmission', 'Color']]
y = df[['Gender']]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2)

model_test = DecisionTreeClassifier()
model_test.fit(X_train, y_train)

y_pred = model_test.predict(X_test)
acc = accuracy_score(y_test, y_pred)
acc
```

```python
import joblib
model = DecisionTreeClassifier()
model.fit(X, y)
joblib.dump(model, 'gender_pred_model.pkl')
```
