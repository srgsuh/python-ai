```python
import time
from IPython.display import clear_output, display

img4 = np.full((30, 30), 255, dtype=np.uint8)
for t in range(20):
    for r in range(30):
        for c in range(30):
            img4[r, c] = 0 if np.random.randint(0, 2) else 220
    plt.imshow(img4, cmap='gray', vmin=0, vmax=255)
    display(plt.gcf())
    clear_output(wait=True)
    time.sleep(1)
    
```
