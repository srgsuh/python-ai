```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from game_of_life import game_of_life
import time
from IPython.display import clear_output, display
```

```python
plt.axis("off")
game = game_of_life(30, 0, 220)
for tick, img in enumerate(game):
    clear_output(wait=True)
    plt.imshow(img, cmap='gray', vmin=0, vmax=255)
    plt.title(f"Game of life. Tick: {tick}")
    display(plt.gcf())
    time.sleep(0.5)
```
