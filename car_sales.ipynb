```python
import pandas as pd
df = pd.read_csv('car_sales_cleaned1.csv')
df
```

```python
df.describe()
```

```python
df = df.drop_duplicates()
df.shape
```

```python
df = df.dropna() # drop rows with null values
df.shape
```

```python
company_values = df['Company'].unique()
company_values
```

```python
df_proj = df[['Annual Income','Price ($)']]
df_proj
```

```python
from sklearn.tree import DecisionTreeClassifier
model_price_income = DecisionTreeClassifier()
x = df_proj[['Price ($)']]
y = df_proj[['Annual Income']]
model_price_income.fit(y, x)
predictions = model_price_income.predict([[31500]])
predictions
```
