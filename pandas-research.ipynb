```python
import pandas as pd
import joblib as jl
df: pd.DataFrame = jl.load('car_sales_data.pkl')
df
```

```python
# Getting Data Frame containing only data about Hyundai
df_hyundai = df.query('Company == "Hyundai"')
df_hyundai
```

```python
# Getting Data Frame containing only data about Toyota with price greater than 40000
min_price = 40000
df_filtered = df.query('Company == "Toyota" and `Price ($)` > @min_price')
df_filtered
```
