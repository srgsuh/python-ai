```python
import pandas as pd
df = pd.read_csv('car_sales_cleaned1.csv')
df
```

```python
df.describe()
```

```python
df = df.drop_duplicates()
df.shape
```

```python
df = df.dropna() # drop rows with null values
df.shape
```

```python
company_values = df['Company'].unique()
company_values
```

```python
df_proj = df[['Annual Income','Price ($)']]
df_proj
```

```python
from sklearn.tree import DecisionTreeClassifier
model_price_income = DecisionTreeClassifier()
y = df_proj[['Price ($)']]
X = df_proj[['Annual Income']]
model_price_income.fit(X, y)
predictions = model_price_income.predict([[30000]])
predictions
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2)

model_split = DecisionTreeClassifier()
model_split.fit(X_train, y_train)

predictions = model_split.predict(X_test)
accuracy_score(y_test, predictions)
```

```python
import converter as conv
df_projected = df[['Company', 'Model']]

mapper = conv.columnsMapper(['Company', 'Model'], df_projected)
df_converted = conv.convertX(df_projected, mapper)

X = df_converted[['Model']]
y = df_projected[['Company']]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)

predictions = model.predict(X_test)
accuracy_score(y_test, predictions)
```
