```python
import pandas as pd
df = pd.read_csv('car_sales_cleaned1.csv').drop_duplicates().dropna()
df.shape
```

```python
import converter as conv
mapper = conv.columnsMapper(['Gender', 'Company', 'Model', 'Transmission', 'Color'], df)
df_conv = conv.convertX(df, mapper)
df_conv.shape
```

```python
from sklearn.model_selection import train_test_split
X = df_conv[['Gender', 'Color']]
y = df[['Model']]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

model_test = DecisionTreeClassifier().fit(X_train, y_train)
y_pred = model_test.predict(X_test)

accuracy_score(y_test, y_pred)
```

```python
import joblib as jl
model_persist = DecisionTreeClassifier().fit(X, y)
jl.dump(model_persist, 'model_by_gender_color.pkl')
```
