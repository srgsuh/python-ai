```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
text = "I want to find a job in the IT field. I’m studying programming at Tel-Ran courses. \
The instructors say I need to do my homework if I want to get hired. \
But because of family responsibilities and other commitments, I can’t find the time for it."
```

```python
text_arr = text.split(".")
text_arr
```

```python
text_clean = [res for s in text_arr if (res := s.strip())]
text_clean
```

```python
query = ["What does instructors say?"]
```

```python
vectorizer = TfidfVectorizer()
```

```python
embeddings = vectorizer.fit_transform(text_clean + query).toarray()
embeddings.shape
```

```python
emb_text = embeddings[:-1]
emb_query = embeddings[-1:]
```

```python
similarities = cosine_similarity(emb_text, emb_query)
similarities
```

```python
best_index = np.argmax(similarities)
text_clean[best_index]
```
