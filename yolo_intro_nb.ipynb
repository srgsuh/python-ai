```python
from ultralytics import YOLO
import matplotlib.pyplot as plt

model = YOLO("yolov8l.pt")
```

```python
results = model("img.jpg")
img_annotated = results[0].plot()

plt.figure(figsize=(8,8))
plt.imshow(img_annotated)
plt.title("Objects detected")
plt.axis("off")
plt.show()
```

```python
model_seg = YOLO("yolov8m-seg.pt")
results = model_seg("img.jpg")
img_seg_annotated = results[0].plot()

plt.figure(figsize=(10,8))
plt.imshow(img_seg_annotated)
plt.title("Objects segmentation")
plt.axis("off")
plt.show()
```

```python
model_seg = YOLO("yolov8l-cls.pt")
results = model_seg("img.jpg")
img_seg_annotated = results[0].plot()

plt.figure(figsize=(5,5))
plt.imshow(img_seg_annotated)
plt.title("Objects classification")
plt.axis("off")
plt.show()
```
